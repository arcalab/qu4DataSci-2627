```python
import pennylane as qml
```

```python
n_qubits = 1

#criar um dispositivo quantico -> um simulador quantico no nosso computador classico. 
dev = qml.device("default.qubit", wires=n_qubits)

#criamos um dispositivo com um qubit. Este qubit começa SEMPRE no estado |0>.

@qml.qnode(dev)
def LCD():
    qml.PauliX(wires=0)
    
    #A função tem que retornar qqr coisa!! 
    return qml.probs(wires=0)
```

```python
LCD()
```

```python
#desenhar o circuito acima
qml.draw_mpl(LCD)() #-> queremos fazer o desenho do circuito que nao recebe argumentos
```
