```python
n_qubits = 2

dev = qml.device("default.qubit", wires=n_qubits)

# o qubit começa sempre em |0>
@qml.qnode(dev)
def qc():
    qml.Hadamard(wires=0)
    qml.PauliZ(wires=0)
    qml.Hadamard(wires=0)

    return qml.probs(wires=0)


qc()



```

How to compare quantum operations in pennylane using their matrix decomposition

```python

dev = qml.device("default.qubit", wires=1)


def circuit():  
    qml.Hadamard(wires=0)
    qml.PauliZ(wires=0)
    qml.Hadamard(wires=0)

def circuit2():  
    qml.PauliX(wires=0)

# Get dense matrix representation for a specific parameter value
matrix_fn = qml.matrix(circuit, [0])
matrix_fn2 = qml.matrix(circuit2, [0])

mat = matrix_fn()
mat2 = matrix_fn2()

print(mat)
print(mat2)

#compare the two matrices
print("Are the two matrices equal?", np.allclose(mat, mat2))
```
