```python
import pennylane as qml
```

```python
n_qubits = 1


dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev)
def qc():
    qml.Hadamard(wires=0)

    return qml.probs(wires=0)
```

```python
qc()
```

```python
n_qubits = 20


dev = qml.device("default.qubit", wires=n_qubits, shots=1000000)

@qml.qnode(dev)
def qc():
    #o qubit começa sempre no estado |0>

    for i in range(n_qubits):
        qml.Hadamard(wires=i)


    
    return qml.probs()

qc()
```

```python
n_qubits = 5


dev = qml.device("default.qubit", wires=n_qubits, shots=1000000)

@qml.qnode(dev)
def qc():
    #o qubit começa sempre no estado |0>

    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)

    
    return qml.probs()

qc()
```

```python
n_qubits = 1


dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev)
def qc(angle):
    #o qubit começa sempre no estado |0>

    qml.RY(angle, wires=0)

    
    return qml.state()

qc(np.pi)
```
